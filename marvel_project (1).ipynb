FASE 2. Trabajo con Datos

```python
class Character:
    def __init__(self, name, publisher, alignment, power):
        self.name = name
        self.publisher = publisher
        self.alignment = alignment
        self.power = power

    def __str__(self):
        return f"{self.name} | {self.publisher} | {self.alignment} | Power: {self.power}"
```

```python
characters = []

for p in personajes_raw:
    char = Character(
        p["name"],
        p["publisher"],
        p["alignment"],
        p["power"]
    )
    characters.append(char)
```

```python
print(characters[0])
```

```python
def buscar_personaje(personajes, texto):
    return [p for p in personajes if texto.lower() in p.name.lower()]
```

```python
resultados = buscar_personaje(characters, "man")
for r in resultados[:5]:
    print(r)
```
